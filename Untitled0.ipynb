```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, Subset
import numpy as np
```

```python

# ======================================================
# 1. CONFIGURATION
# ======================================================

torch.manual_seed(42)
np.random.seed(42)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

BATCH_SIZE = 64
EPOCHS = 5
LEARNING_RATE = 0.001
```

```python

# ======================================================
# 2. DATASET MNIST : CHIFFRES DE 0 À 9 (28x28)
# ======================================================

transform = transforms.Compose([
    transforms.ToTensor()
])

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

# Conversion en NumPy
X_train = train_dataset.data.numpy().reshape(-1, 1, 28, 28).astype(np.float32)
y_train = train_dataset.targets.numpy().astype(np.int64)

X_test = test_dataset.data.numpy().reshape(-1, 1, 28, 28).astype(np.float32)
y_test = test_dataset.targets.numpy().astype(np.int64)

# Normalisation des pixels entre 0 et 1
# Identique à transforms.ToTensor()
X_train = X_train / 255.0
X_test = X_test / 255.0

print(f'Train set: {X_train.shape[0]} images')
print(f'Test set:  {X_test.shape[0]} images')
```

```python

# ======================================================
# 3. ARCHITECTURE DU PETIT CNN
# ======================================================

class TinyCNN(nn.Module):

    def __init__(self):
        super().__init__()

        # Entrée : 1 x 28 x 28
        # Sortie : 4 x 26 x 26
        self.conv = nn.Conv2d(
            in_channels=1,
            out_channels=4,
            kernel_size=3,
            stride=1,
            padding=0
        )

        self.relu = nn.ReLU()

        # Sortie : 4 x 13 x 13
        self.pool = nn.MaxPool2d(
            kernel_size=2,
            stride=2
        )

        # 4 * 13 * 13 = 676 entrées
        self.fc = nn.Linear(676, 10)

    def forward(self, x):

        x = self.conv(x)
        x = self.relu(x)
        x = self.pool(x)

        x = torch.flatten(x, start_dim=1)

        x = self.fc(x)

        return x

model = TinyCNN().to(device)

print(model)
print(
    "Nombre de paramètres :",
    sum(p.numel() for p in model.parameters())
)
```

```python


# ======================================================
# 4. ENTRAÎNEMENT
# ======================================================

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

for epoch in range(EPOCHS):

    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

        predictions = outputs.argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    accuracy = 100 * correct / total

    print(
        f"Epoch {epoch+1}/{EPOCHS} "
        f"| Loss = {total_loss/len(train_loader):.4f} "
        f"| Accuracy train = {accuracy:.2f}%"
    )
```

```python


# ======================================================
# 5. ÉVALUATION SUR LE TEST SET
# ======================================================

model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        predictions = outputs.argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

test_accuracy = 100 * correct / total

print(f"\nAccuracy test FP32 : {test_accuracy:.2f}%")

# ======================================================
# 6. SAUVEGARDE DU MODÈLE FP32
# ======================================================

torch.save(
    model.cpu().state_dict(),
    "tiny_cnn_fp32.pth"
)

# ======================================================
# 7. EXPORT DES POIDS POUR LE FPGA
# ======================================================

conv_weights = model.conv.weight.detach().numpy()
conv_bias = model.conv.bias.detach().numpy()

fc_weights = model.fc.weight.detach().numpy()
fc_bias = model.fc.bias.detach().numpy()

np.savez(
    "tiny_cnn_weights_fp32.npz",
    conv_weights=conv_weights,
    conv_bias=conv_bias,
    fc_weights=fc_weights,
    fc_bias=fc_bias
)

print("\nPoids exportés avec succès")

print("Conv weights :", conv_weights.shape)
print("Conv bias    :", conv_bias.shape)
print("Dense weights :", fc_weights.shape)
print("Dense bias    :", fc_bias.shape)
```

```python
W = dict(np.load("tiny_cnn_weights_fp32.npz"))
for k, v in W.items():
    print(f'  {k}: {v.shape}')
```

```python
def np_flatten(x):
    return x.reshape(x.shape[0], -1)

# STUDENT TASK: implement np_linear and np_relu
def np_linear(x, weight, bias):
    # BEGIN
    return x@weight.T + bias
    # END


def np_conv2d(x, weight, bias):
    B, C_in, H, W = x.shape
    C_out, _, K, _ = weight.shape

    out = np.zeros((B, C_out, H-K+1, W-K+1),dtype=np.result_type(x, weight, bias))

    for c in range(C_out):
        for i in range(H-K+1):
            for j in range(W-K+1):
                window = x[:, :, i:i+K, j:j+K]
                out[:, c, i, j] = np.sum(
                    window * weight[c],
                    axis=(1, 2, 3)
                ) + bias[c]

    return out



def np_maxpool(x):
    B, channels, H, W = x.shape

    out = np.zeros((B, channels, H//2, W//2), dtype=x.dtype)

    for i in range(H//2):
        for j in range(W//2):
            window = x[:, :, 2*i:2*i+2, 2*j:2*j+2]
            out[:, :, i, j] = np.max(window, axis=(2, 3))

    return out


def np_relu(x):
    # STUDENT TASK: implement ReLU
    # BEGIN
    return np.maximum(0,x)
    # END
def get_scale(x, bits):
    # BEGIN
    scale = np.max(np.abs(x))/(2**(bits-1)-1)
    # END
    return scale

def quantize(x, scale):
    # BEGIN
    x_q = np.round(x / scale)
    # END
    return x_q.astype(np.int64)
```

```python
def calibrate(X_cal, W, bits_a, num_samples):
    act_scales = {}
    act_max = {}
    act_max_img = []
    # Activation scales
    for x in X_cal[:num_samples]:
  # STUDENT TASK: implement the forward pass of the model using np_linear and np_relu,
        # and record the maximum absolute value of each layer's input
        x = x[None, ...]
        act_max_layers = []

        act_max_layers.append(np.max(np.abs(x)))# (1, 28, 28)
        x = np_relu(np_conv2d(x, W['conv_weights'], W['conv_bias'])) # (4, 26, 26)
        x = np_maxpool(x) # (4, 13, 13)
        act_max_layers.append(np.max(np.abs(x)))
        x = x.flatten() # (676)
        x = np_linear(x, W['fc_weights'], W['fc_bias']) # (10)
        act_max_img.append(act_max_layers)

    max_act_rows = np.max(act_max_img,axis = 0)


    act_max['0'] = max_act_rows[0]
    act_max['1'] = max_act_rows[1]
    act_scales['0'] = act_max['0']/(2**(bits_a-1)-1)
    act_scales['1'] = act_max['1']/(2**(bits_a-1)-1)

    return act_scales

act_scales = calibrate(X_train, W, bits_a=bits, num_samples=1000)
k_act0 = int(np.ceil(np.log2(act_scales['0']))) if act_scales['0'] > 0 else 0
act_scale0_power2 = 2**k_act0

def quantize_model(W, bits_w, act_scales):
    W_q = {}
    out_scales = {}

    W_q['conv_weights'] = quantize(W['conv_weights'], get_scale(W['conv_weights'], bits_w))
    out_scales['0'] = act_scale0_power2 * get_scale(W['conv_weights'], bits_w)
    W_q['conv_bias'] = quantize( W['conv_bias'] ,  out_scales['0'])

    W_q['fc_weights'] = quantize(W['fc_weights'], get_scale(W['fc_weights'], bits_w))
    out_scales['1'] = act_scales['1'] * get_scale(W['fc_weights'], bits_w)
    W_q['fc_bias'] = quantize( W['fc_bias'] ,  out_scales['1'])

    return W_q, out_scales

```

```python
bits = 8

# Calibrate scales and quantize weights
act_scales = calibrate(X_train, W, bits_a=bits, num_samples=1000)
W_q, out_scales = quantize_model(W, bits_w=bits, act_scales=act_scales)

# TASK: compute merged scales between consecutive layers
merged_scales = {}
merged_scale_power2 = {}
# BEGIN
merged_scales['0'] = out_scales['0']/act_scales['1']
k = int(np.ceil(np.log2(merged_scales['0']))) if merged_scales['0'] > 0 else 0
merged_scale_power2['0'] = 2**k
k_act0 = int(np.ceil(np.log2(act_scales['0']))) if act_scales['0'] > 0 else 0
act_scale0_power2 = 2**k_act0

def np_linear_merged(x_q, w_q, b_q, merged_scale, bits_a):
    # STUDENT TASK: implement quantized linear with merged scales.
    # Don't forget to clip the output to the range of int8.
    # BEGIN
    out_q = x_q @ w_q.T + b_q
    out_q = np.round(out_q * merged_scale)
    qmax = (2**(bits_a-1)-1)
    out_q = np.clip(out_q ,-qmax , qmax)

    # END
    return out_q

def np_model_merged(x, W_q, merged_scales, bits):
    # STUDENT TASK: implement the forward pass of the model using np_linear_merged and np_relu
    # Don't forget to quantize the input using input_scale, and handle the last layer
    # BEGIN
    qmax = (2**(bits-1)-1)


    x = np.clip(quantize(x, act_scale0_power2),-qmax, qmax).astype(np.int64)
    x = np_conv2d(x, W_q['conv_weights'], W_q['conv_bias']) # (4, 26, 26)
    x = np.round(x >> np.abs(k))
    x = np.clip(x, -qmax, qmax).astype(np.int64)
    x = np_relu(x)
    x = np_maxpool(x) # (4, 13, 13)
    x = x.reshape(x.shape[0], -1) # (676)
    x = np_linear(x, W_q['fc_weights'], W_q['fc_bias']) # (10)

    return x

preds_m = np_model_merged(X_test, W_q, merged_scale_power2, bits).argmax(axis=1)
acc_m = (preds_m == y_test).mean()
print(f'Merged scale int8: {acc_m*100:.2f}%')
```
